# Representação de textos: BoW e TF-IDF

## 1. Corpus
Dez frases curtas.

```python
corpus = [
    "o gato dorme no sofa",
    "o cachorro dorme no quintal",
    "o gato come peixe",
    "o cachorro come carne",
    "eu gosto de gato",
    "eu gosto de cachorro",
    "o peixe nada no rio",
    "a carne esta na geladeira",
    "o sofa esta na sala",
    "o rio esta cheio de peixe",
]
```

## 2. Pré-processamento
Separa cada texto em palavras e monta o vocabulário (todas as palavras diferentes, em ordem alfabética).

```python
textos = [texto.split() for texto in corpus]
vocabulario = sorted(set(palavra for texto in textos for palavra in texto))
print(vocabulario)
```

## 3. Bag of Words (BoW)
Para cada texto, conta quantas vezes cada palavra do vocabulário aparece.

```python
bow = [[texto.count(palavra) for palavra in vocabulario] for texto in textos]

for i, vetor in enumerate(bow):
    print(f"Texto {i+1}: {vetor}")
```

## 4. TF-IDF
- **TF** = vezes que a palavra aparece no texto / total de palavras do texto
- **IDF** = log(N / número de textos que contêm a palavra)
- **TF-IDF** = TF × IDF

Palavras que aparecem em muitos textos (ex.: "o") ficam com peso baixo; palavras raras ficam com peso alto.

```python
import math

N = len(textos)

def tf(palavra, texto):
    return texto.count(palavra) / len(texto)

def idf(palavra):
    df = sum(1 for texto in textos if palavra in texto)
    return math.log(N / df)

tfidf = [[round(tf(p, texto) * idf(p), 3) for p in vocabulario] for texto in textos]

for i, vetor in enumerate(tfidf):
    print(f"Texto {i+1}: {vetor}")
```
